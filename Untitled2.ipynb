```python
import pandas as pd
df=pd.read_csv('/content/1776250375-P2-Healthcare Analytics.csv')
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
df['gender'].value_counts()
```

```python
import matplotlib.pyplot as plt
df['gender'].value_counts().plot(kind='bar')
plt.title('pateint count by gender')
plt.xlabel('gender')
plt.ylabel('number of pateints')
plt.show()
```

```python
df['visits'].value_counts().sort_index()
```

```python
df['visits'].value_counts().sort_index().plot(kind='bar')
plt.title('distribution of doctor visits')
plt.xlabel('no. of visits')
plt.ylabel('number of records')
plt.show()
```

```python
df.groupby('gender')['visits'].mean()
```

```python
df.groupby('gender')['visits'].mean().plot(kind='bar')
plt.title('average visits by gender')
plt.xlabel('gender')
plt.ylabel('average visits')
plt.show()
```

```python
df.groupby('age')['visits'].mean()
```

```python
df.groupby('age')['visits'].mean().plot(kind='line')
plt.title('average visits by age')
plt.xlabel('age')
plt.ylabel('average visits')
plt.show()
```

```python
df.groupby('illness')['visits'].mean()
```

```python
df.groupby('illness')['visits'].mean().plot(kind='bar')
plt.title('average visits by illness')
plt.xlabel('illness')
plt.ylabel('average visits')
plt.show()
```
